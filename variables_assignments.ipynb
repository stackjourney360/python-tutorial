# What is a Variable?
### "A variable is like a container or a name that stores a value in memory, so we can reuse it."

```python
i=10
print(i)
```

## Rules for Variables
    
    • a-z, A-Z, 0-9 and '_' allowd.
    • Start with letters/underscore, not numbers.
    • Case-sensitive (age vs Age).
    • No special characters except _.
    • Avoid reserved keywords (like class, def).

```python
i=10
j2=11
_iI=12
_iI
#9i=13 invalid
```

# Data Types in Python


### (a) Numeric Types

	1) int – integers 
	2) float – decimals 
	3) complex – complex numbers 

```python
i=10
print(i, type(i))
x=10.5
print(x, type(x))
c=2+3j
print(c, type(c))
```

### (b) Text Type
       1) str (String)

```python
s="sajid"
print(s, type(s))
```

### (c) Sequence Types
        1) list 
        2) tuple
        3) range 

```python
l=[1,2,3]
print(type(l))
```

```python
t=(1,2,3)
print(type(t))
```

```python
r=range(5)
print(r, type(r))
```

### (d) Mapping Type
        1) dict

```python
d={"name":"Sajid","Marks":"200"}
print(d, type(d))
```

### (e) Set Types
        1) set
        2) frozenset

```python
s={1,2,3,1,2,3}
s, type(s)
```

```python
l=[1,2,3,4,4]
f=frozenset(l)
print(f, type(f))
```

### (e) Boolean Type
        1) bool
      

```python
b=False
print(b, type(b))
```

# Variable Assignment 
### 1. Simple Assignment
### 2. Multiple Variables to Same Value
### 3. Multiple Assignments in One Line
### 4. Unpacking with Lists or Tuples
### 5. Extended Unpacking

```python
# Simple Assignment
i=10
print(i)
```

```python
# Multiple Variables to Same Value
x=z=i=[1,2,3]
print(x,z,i)
```

```python
# Multiple Assignments in One Line
i,j,c=1,2,3
print(i,j,c)
```

```python
# Unpacking with Lists or Tuples
l=[1,2,3]
x,y,z=l
print(x,y,z)
```

```python
# Extended Unpacking
a, *b, c=[1,2,3,4,5,6]
print(a,b,c)
```

## Type Casting (Conversion)
